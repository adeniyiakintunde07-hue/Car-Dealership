```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

teams = ["Arsenal", "Man City", "Liverpool", "Chelsea", "Aston Villa"]
wins = np.array([25, 27, 24, 18, 20])
draws = np.array([5, 4, 6, 8, 7])
losses = np.array([8, 7, 8, 12, 11])

points = wins * 3 + draws

df = pd.DataFrame({
    "Team": teams,
    "Wins": wins,
    "Draws": draws,
    "Losses": losses,
    "Points": points
})

print(df)

df = df.sort_values("Points", ascending=False)
print("\nSorted Table:")
print(df)

plt.bar(df["Team"], df["Points"], color="purple")
plt.title("Football Team Points")
plt.xlabel("Team")
plt.ylabel("Points")
plt.show()
```
