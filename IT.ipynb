```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

songs = ["Calm Down", "Rush", "Unavailable", "Soweto", "Bandana"]
artists = ["Rema", "Ayra Starr", "Davido", "Victony", "Fireboy"]
streams = np.array([95, 90, 88, 70, 85])
likes = np.array([80, 75, 78, 60, 72])
ratings = np.array([10, 9, 9, 8, 9])

print("Average streams:", streams.mean())
print("Highest stream:", streams.max())
print("Lowest stream:", streams.mean())
print("Average likes:", likes.mean())
print("Average rating", ratings.mean())

popularity_score = streams + likes
print("Popularity score:", popularity_score)

#Pandas
data = {
    "Song": songs,
    "Artist": artists,
    "Streams": streams,
    "Likes": likes,
    "Rating": ratings
}
df = pd.DataFrame(data)
print(df)

print("\n")
df["Popularity Score"] = df["Streams"] + df["Likes"]
print(df)

print("\n")
#Sorting
df = df.sort_values("Popularity Score", ascending=False)
print("Sort by Popularity Score")
print(df)

print("\n")
#Filtering
top_songs = df[df["Streams"] > 85]
print("Filter by Top Songs")
print(top_songs)

#Bar Chart for Streams
plt.bar(df["Song"], df["Streams"], color="green")
plt.title("Afrobeats Song Streams")
plt.xlabel("Song")
plt.ylabel("Streams")
plt.show()

#Bar Chart for Popularity Score
plt.bar(df["Song"], df["Popularity Score"], color="orange")
plt.title("Song Popularity")
plt.xlabel("Song")
plt.ylabel("Popularity Score")
plt.legend()
plt.show()

#Line Chart for Rating
plt.plot(df["Song"], df["Rating"], marker="o", color="blue")
plt.title("Song Popularity Score")
plt.xlabel("Song")
plt.ylabel("Rating")
plt.show()
```
