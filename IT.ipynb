```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]
lagos_temp = np.array([31, 30, 32, 31, 33])
london_temp = np.array([18, 17, 19, 16, 18])

#Step 2: Numpy Calculation
print("\nNumpy Calculation")
print("Average Lagos temperature: ", lagos_temp.mean())
print("Average London temperature: ", london_temp.mean())
print("Highest Lagos temperature: ", lagos_temp.max())
print("Lowest Lagos temperature: ", london_temp.min())

difference = lagos_temp - london_temp
print("Temperature difference each day: ", difference)

# Step 3: Create a DataFrame with Pandas
data = {
    "Day": days,
    "Lagos Temp": lagos_temp,
    "London Temp": london_temp
}

df = pd.DataFrame(data)

# Step 4: Add new column
df["Difference"] = df["Lagos Temp"] - df["London Temp"]

print("\nPandas DataFreame")
print(df)

# Step 5: Filter
warm_days = df["Lagos Temp"] > 31
print("\nLagos days above 31 degrees:")
print(warm_days)

# Step 6: Create a Chart with Matplotlib
plt.plot(df["Day"], df["Lagos Temp"], marker="o", color="red", label="Lagos")
plt.plot(df["Day"], df["London Temp"], marker="o", color="blue", label="London")
plt.title("Lagos vs London Temperature")
plt.xlabel("Days")
plt.ylabel("Temperature")
plt.legend()
plt.show()

plt.bar(df["Day"], df["Difference"], color="purple")
plt.title("Temperature Difference: Lagos - London")
plt.xlabel("Days")
plt.ylabel("Difference")
plt.show()


```
