```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

players = ["Saka", "Haaland", "Salah", "Palmer", "Watkins"]
goals = np.array([15, 27, 18, 20, 19])
assists = np.array([10, 5, 9, 8, 7])

#Use Numpy for Calculations
print("Average goals: ", goals.mean())
print("Highest goals: ", goals.max())
print("Lowest goals: ", goals. min())
print("Average Assists: ", assists.mean())

contribution = goals + assists
print("Total contribution: ", contribution)

#Use Pandas to Create a DataFrame (Table)
data = {
    "Player": players,
    "Goals": goals,
    "Assists": assists
}

df = pd.DataFrame(data)
print(df)

print("\n")
df["Total Contribution"] = df["Goals"] + df["Assists"]
print(df)

df = df.sort_values("Total Contribution", ascending=False)
print(df)

#Filtering
top_players = df[df["Goals"] > 18]
print(top_players)

#Creating Charts
#Bar Charts for Goals
plt.bar(df["Player"], df["Goals"], color="green")
plt.title("Player Goals")
plt.xlabel("Player")
plt.ylabel("Goals")
plt.show()

#Bar Chart for Total Contribution
plt.bar(df["Player"], df["Total Contribution"], color="blue")
plt.title("Total Contribution")
plt.xlabel("Player")
plt.ylabel("Goals + Assists")
plt.show()

print("\nInsight: Haaland scored the highest number of goals.")
```
